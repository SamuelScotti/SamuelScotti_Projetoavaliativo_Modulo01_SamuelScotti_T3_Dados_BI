## 1. Importação das Bibliotecas e Dados

Nesta etapa, importei a biblioteca Pandas e carreguei os ficheiros CSV gerados pelas consultas SQL no FreeSQL.

```python
import pandas as pd

# Carregar os dados extraídos das consultas SQL
df_query1 = pd.read_csv('query_01.csv')
df_query2 = pd.read_csv('query_02.csv')

# Exibir as primeiras 5 linhas da Query 1 para confirmar a importação
display(df_query1.head())
```

## 2. Estatísticas Descritivas de Remuneração

Cálculo da média, mediana e valores extremos para entender a distribuição de salários na empresa.

```python
# Calcular medidas estatísticas básicas do Salário (Query 1)
media_salarial = df_query1['SALARY'].mean()
mediana_salarial = df_query1['SALARY'].median()
salario_minimo = df_query1['SALARY'].min()
salario_maximo = df_query1['SALARY'].max()

print("--- Estatísticas Salariais ---")
print(f"Média: $ {media_salarial:.2f}")
print(f"Mediana: $ {mediana_salarial:.2f}")
print(f"Mínimo: $ {salario_minimo:.2f}")
print(f"Máximo: $ {salario_maximo:.2f}")
```

## 3. Visualização Gráfica

Geração de gráficos para visualizar com facilidade a concentração dos salários e a disparidade entre diferentes departamentos.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Configurações de estilo visual para os gráficos
sns.set_theme(style="whitegrid")
plt.figure(figsize=(14, 6))

# 1. Histograma da Distribuição Salarial
plt.subplot(1, 2, 1)
sns.histplot(df_query1['SALARY'], bins=10, kde=True, color='skyblue')
plt.title('Distribuição de Salários')
plt.xlabel('Salário (USD)')
plt.ylabel('Frequência (Número de Colaboradores)')

# 2. Boxplot de Salários por Departamento
plt.subplot(1, 2, 2)
sns.boxplot(x='SALARY', y='DEPARTMENT_NAME', data=df_query1, hue='DEPARTMENT_NAME', palette='Set2', legend=False)
plt.title('Distribuição de Salários por Departamento')
plt.xlabel('Salário (USD)')
plt.ylabel('Departamento')

plt.tight_layout()
plt.show()
```
